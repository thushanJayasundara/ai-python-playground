#### Exercise 6.2

This exercise is related to exercise 5.2 of the previous week. Consider the data from CSV file `weight-height.csv`.

1) Read data into a pandas dataframe.

2) Pick the target variable `y` as weight in kilograms, and the feature variable `X` as height in centimeters.

3) Split the data into training and testing sets with 80/20 ratio.

4) Scale the training and testing data using normalization and standardization.

4) Fit a KNN regression model with `k=5` to the training data without scaling, predict on unscaled testing data and compute the $R^2$ value.

6) Repeat step 4 for normalized data.

7) Repeat step 4 for standardize data.

8) Compare the models in terms of their $R^2$ value.

```python
#### Exercise 6.1

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, classification_report

# 1) Read dataset
df = pd.read_csv("data_banknote_authentication.csv")

# 2) Features and target
X = df.drop("class", axis=1)
y = df["class"]

# 3) Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=20
)

# 4) Linear SVM
svc_linear = SVC(kernel="linear")
svc_linear.fit(X_train, y_train)
y_pred_linear = svc_linear.predict(X_test)

print("Linear Kernel – Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_linear))
print("\nLinear Kernel – Classification Report:")
print(classification_report(y_test, y_pred_linear))

# 6) RBF SVM
svc_rbf = SVC(kernel="rbf")
svc_rbf.fit(X_train, y_train)
y_pred_rbf = svc_rbf.predict(X_test)

print("RBF Kernel – Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_rbf))
print("\nRBF Kernel – Classification Report:")
print(classification_report(y_test, y_pred_rbf))

# 7) Summary
print("""
Comparison Summary:
- The Linear SVM works well when the data is close to linearly separable. It is
  simpler to train and usually faster.
- The RBF SVM can model nonlinear patterns, so it is more flexible and can capture
  more complex relationships.
- In this dataset, both models perform well, but RBF may have a slight advantage
  if the decision boundary is not perfectly linear.
""")
```

```python
#### Exercise 6.2

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.metrics import r2_score

df = pd.read_csv("weight-height.csv")

X = df[["Height"]]
y = df["Weight"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=20
)

knn_unscaled = KNeighborsRegressor(n_neighbors=5)
knn_unscaled.fit(X_train, y_train)
y_pred_unscaled = knn_unscaled.predict(X_test)
r2_unscaled = r2_score(y_test, y_pred_unscaled)

norm = MinMaxScaler()
X_train_norm = norm.fit_transform(X_train)
X_test_norm = norm.transform(X_test)

knn_norm = KNeighborsRegressor(n_neighbors=5)
knn_norm.fit(X_train_norm, y_train)
y_pred_norm = knn_norm.predict(X_test_norm)
r2_norm = r2_score(y_test, y_pred_norm)

std = StandardScaler()
X_train_std = std.fit_transform(X_train)
X_test_std = std.transform(X_test)

knn_std = KNeighborsRegressor(n_neighbors=5)
knn_std.fit(X_train_std, y_train)
y_pred_std = knn_std.predict(X_test_std)
r2_std = r2_score(y_test, y_pred_std)

print("R2 Unscaled:", r2_unscaled)
print("R2 Normalized:", r2_norm)
print("R2 Standardized:", r2_std)

print("""
Comparison Summary:
- The unscaled KNN model is influenced by the raw height values because KNN
  relies on distance, and larger-scale features dominate the calculation.
- After applying normalization or standardization, the model performs better
  because scaling ensures that distances are measured fairly.
- As a result, the normalized and standardized versions achieve higher R² scores.
""")
```
